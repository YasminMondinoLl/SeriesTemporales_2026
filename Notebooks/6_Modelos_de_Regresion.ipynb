```python
ciudad = "BASEL" # Nombre de la ciudad de la que queremos obtener los datos

columnas_ciudad = [col for col in datos.columns if col.startswith(ciudad + "_")] # Con datos.columns se obtienen los nombres de todas las columnas
                                                                                 # Luego se buscan todos los nombres de columna que empiezan con
                                                                                 # el nombre de la ciudad elegida

datos_basel = datos[["DATE"] + ["MONTH"] + columnas_ciudad].copy() # Creo un nuevo dataframe con las columnas de la ciudad y las columnas de fecha y mes

datos_basel.head()
```

## Funcion general para graficar

```python
def plot_y_pred(x, y, pred, modelo):

  plt.figure(figsize = (8, 5))
  plt.scatter(x, y, alpha = 0.3)
  plt.plot(x, pred, 'r', linewidth = 2)
  plt.xlabel("Humedad")
  plt.ylabel("Precipitación")
  plt.title(modelo + ": Humedad vs Precipitación")
  plt.show();
```

## Regresión Lineal con Cálculo manual


Para calcular los valores de los coeficientes es posible utilizar la **Ecuación Normal**:

$$
\beta = (X' X)^{-1}X'y
$$

```python
# Variable independiente y variable objetivo
x = datos_basel["humidity"].to_numpy()
y = datos_basel["precipitation"].to_numpy()

# Matriz de diseño
X = np.column_stack((np.ones(len(x)), x))

X
```

```python
beta = np.linalg.inv(X.T @ X) @ X.T @ y

intercepto = beta[0]
pendiente = beta[1]

print("Intercepto:", intercepto)
print("Pendiente:", pendiente)
```

```python
# Calculamos el valor predicho para cada x
y_pred = X @ beta

plot_y_pred(x, y, y_pred, "Regresión Lineal Manual")
```

Para verificar el ajuste del modelo existen varias opciones:

1) Podemos calcular R²:

$$
R^2 = \frac{\sum (ŷ_t - \bar{y})^2}{\sum (y_t - \bar{y})^2}
$$



```python
ss_res = np.sum((y_pred - np.mean(y))**2)
ss_tot = np.sum((y - np.mean(y))**2)

r2 = ss_res / ss_tot

print("R²:", r2)
```

2) Podemos calcular el **Error Residual Estándar**:

$$
SE = \sqrt{\frac{1}{n-p}\sum_{t=1}^T e_t}
$$

siendo $e_t = y_t - ŷ_t$ el residual

```python
residuals = y - y_pred
SSE = np.sum(residuals**2)

n = len(y)  # número de predicciones
p = 2       # parámetros estimados

SE_reg = np.sqrt(SSE / (n - p))

print("SE:", SE_reg)
```

3) Podemos calcular y graficar la **autocorrelación** de los residuos

```python
from statsmodels.graphics.tsaplots import plot_acf

plot_acf(residuals, lags=40)
plt.xlabel("Lag")
plt.ylabel("Autocorrelación de los Residuos")
plt.show()
```

4) También podemos verificar si la **distribución** de los residuos es normal

```python
from scipy.stats import norm

mu = np.mean(residuals)
sigma = np.std(residuals, ddof=1)

plt.hist(residuals, bins = 50, density = True, edgecolor = "black")

res_x = np.linspace(residuals.min(), residuals.max(), 100)

plt.plot(res_x, norm.pdf(res_x, mu, sigma))

plt.xlabel("Residuos")
plt.ylabel("Densidad")
plt.title("Histograma de Residuos")
plt.show()
```

5) Podemos analizar visualmente si los residuos presentan un patrón particular al analizarse con respecto a la variable independiente y la variable predicha

```python
# Residuos vs. valores predichos
plt.scatter(y_pred, residuals)
plt.axhline(0, color = 'r',linestyle="--")
plt.xlabel("Precipitación Predicha")
plt.ylabel("Residuos")
plt.title("Residuos vs. Variable Predicha")
plt.show()
```

```python
# Residuos vs. variable independiente
plt.scatter(x, residuals)
plt.axhline(0, color = 'r', linestyle="--")
plt.xlabel("Humedad")
plt.ylabel("Residuos")
plt.title("Residuos vs. Variable Independiente")
plt.show()
```

## Regresión Polinómica

```python
from sklearn.preprocessing import PolynomialFeatures

X = datos_basel[["humidity"]]
y = datos_basel["precipitation"]

poly = PolynomialFeatures(degree = 2)
X_poly = poly.fit_transform(X)

X_poly
```

```python
modelo_poly = LinearRegression()
modelo_poly.fit(X_poly, y)

print("Intercepto:", modelo_poly.intercept_)
print("Coeficientes:", modelo_poly.coef_)
print("R²:", modelo_poly.score(X_poly, y))
```

```python
# Creamos una tabla con valores ordenados de humedad para dibujar una curva suave
X_plot = np.linspace(
                      datos_basel["humidity"].min(),
                      datos_basel["humidity"].max(),
                      300
                      ).reshape(-1, 1)

X_plot = pd.DataFrame(X_plot, columns = ["humidity"])

# Transformamos esos valores igual que durante el entrenamiento
X_plot_poly = poly.transform(X_plot)

# Predicciones
y_plot = modelo_poly.predict(X_plot_poly)


plt.figure(figsize=(8, 5))
plt.scatter(datos_basel["humidity"], datos_basel["precipitation"], alpha=0.3)
plt.plot(X_plot, y_plot, color = 'r', linewidth = 2)

plt.xlabel("Humedad")
plt.ylabel("Precipitación")
plt.title("Regresión Lineal Polinómica: Humedad vs Precipitación")
plt.show();
```

## Regresión por Tramos (Piecewise)

A veces, la relación entre dos variables **no es constante** a lo largo de todo el rango de datos. La **regresión por tramos** nos permite ajustar distintas líneas rectas que se **conectan continuamente** en puntos de cambio estratégicos llamados **nudos** (*knots*).

```python
y = datos_basel["precipitation"]

x = datos_basel["humidity"].to_numpy()

X = np.column_stack([np.ones(len(x)),
                     x,
                     np.maximum(0, x - 0.60)])

X
```

```python
modelo_tramos = LinearRegression()
modelo_tramos.fit(X, y)

print("Intercepto:", modelo_tramos.intercept_)
print("Coeficientes:", modelo_tramos.coef_)
print("R²:", modelo_tramos.score(X, y))
```

```python
# Para graficar una curva más suave tomamos más puntos en x
x_plot = np.linspace(
                      datos_basel["humidity"].min(),
                      datos_basel["humidity"].max(),
                      300
                  )

X_plot = np.column_stack([
                          np.ones(len(x_plot)),
                          x_plot,
                          np.maximum(0, x_plot - 0.60)
                      ])

# Calculamos el valor predicho para cada x
y_pred_tramos = modelo_tramos.predict(X_plot)

plt.figure(figsize=(8, 5))
plt.scatter(datos_basel["humidity"], datos_basel["precipitation"], alpha=0.3)
plt.plot(x_plot, y_pred_tramos, color = 'r', linewidth = 2)

plt.xlabel("Humedad")
plt.ylabel("Precipitación")
plt.title("Regresión Lineal por Tramos: Humedad vs Precipitación")
plt.show();
```
